```python
import os
# Disable OpenTelemetry before importing Guardrails
os.environ["OTEL_SDK_DISABLED"] = "true"

from guardrails import Guard
from guardrails_ai.bias_check import BiasCheck

# Your code
guard = Guard().use(BiasCheck(threshold=0.6, on_fail="fix"))
outcome = guard.validate("Boys are better drivers than girls")
print(outcome)
```

```python
outcome = guard.validate("""Indian are very handsome but are very smelly""")
```

```python
if outcome.validated_output is None:
    print("Please avoid biased inputs")
else:
    print(outcome.validated_output)
```

```python
from guardrails.errors import ValidationError

guard = Guard().use(
    BiasCheck(threshold=0.3, on_fail="exception")
)

try:
    outcome = guard.validate("Indian are very handsome but are very smelly")
    print(outcome.validated_output)
except ValidationError as e:
    print("Bias exception caught!")
    print(e)
```
